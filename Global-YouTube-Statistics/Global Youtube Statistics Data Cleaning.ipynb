```python
import pandas as pd
import numpy as np
import re
```

```python
# LOADING RAW DATASET (RAW BACKUP)
# This 'df' will remain untouched as raw backup
df = pd.read_csv(
    r"F:\Data Analytics Real time Project portfolio\Global Youtube Statistics\Global YouTube Statistics.csv", 
    encoding='latin-1'
)
```

```python
# Disable scientific notation globally so numbers look clean
pd.set_option('display.float_format', lambda x: '%.0f' % x)
```

```python
# CREATING A WORKSET COPY (DUPLICATE DATASET)
# We make a duplicate copy of 'df' and name it 'df_cleaned'. All cleaning happens here.
df_cleaned = df.copy()
```

```python
# 1. Clear any truncation limits for rows and columns
pd.set_option('display.max_rows', None)
pd.set_option('display.max_columns', None)

# 2. Display the dataframe again
df_cleaned
```

```python
# CLEANING TEXT COLUMNS (REMOVING EMOJIS & NON-ENGLISH CHARACTERS)

# Defining a function that keeps only English alphabets, numbers, and basic punctuation
def clean_text(text):
    if pd.isna(text):
        return text  # If the cell is empty (NaN), leave it as it is
    
    text = str(text)
    # REGEX: Keeps only English letters, numbers, spaces, and standard dashes
    cleaned = re.sub(r'[^a-zA-Z0-9\s\-_.,()]', '', text)
    cleaned = " ".join(cleaned.split())
    return cleaned

# List of text columns we want to convert in English text
text_columns = ['Youtuber', 'category', 'Title', 'Country', 'channel_type']
```

```python
# Applying the cleaning function ONLY to our duplicate dataframe 'df_cleaned'
for col in text_columns:
    if col in df_cleaned.columns:
        df_cleaned[col] = df_cleaned[col].apply(clean_text)

# 1. Clear any truncation limits for rows and columns
pd.set_option('display.max_rows', None)
pd.set_option('display.max_columns', None)

# 2. Display the dataframe again
df_cleaned
```

```python
# Data Cleaning Process

# Step 1. Remove dupliates
# Step 2. Data Standarization
# STep 3. Remove Null values and blank values 
# Step 4. Remove unnecessary columns 

# So we have alot of null values and blank values so we will start from step 3

# STep 3. Remove Null values and blank values 

# 3.1. Replace empty blanks/spaces ('') with 0
df_cleaned = df_cleaned.replace(r'^\s*$', 0, regex=True)

# 3.2. Replace all NaN (Null) values with "Unknown"
df_cleaned = df_cleaned.fillna("Unknown")

# 1. Clear any truncation limits for rows and columns
pd.set_option('display.max_rows', None)
pd.set_option('display.max_columns', None)

# 2. Display the dataframe again
df_cleaned
```

```python
# Step 1. Remove dupliates

# 1.1. Check how many exact duplicate rows exist in df_cleaned
duplicate_count = df_cleaned.duplicated().sum()
print(f"Total duplicate rows found: {duplicate_count}")

# 1.2. Drop duplicates and keep the first occurrence
df_cleaned = df_cleaned.drop_duplicates()

# 1.3. Confirm that duplicates are removed
print(f"Duplicates remaining: {df_cleaned.duplicated().sum()}")
print(f"New dataset shape: {df_cleaned.shape}")
```

```python
# Step 2. Data Standardization

# Function to clean spaces, remove bad special characters, and fix text structure
def standardize_text(text):
    if text == "Unknown" or pd.isna(text):
        return "Unknown"
    
    text = str(text)
    
    # 1. Remove emojis and non-standard symbols but keep letters, numbers, and basic symbols
    cleaned = re.sub(r'[^a-zA-Z0-9\s\-_.,()&]', '', text)
    
    # 2. Fix extra spaces and structural spacing issues (e.g., "Hi   There" -> "Hi There")
    cleaned = " ".join(cleaned.split())
    
    return cleaned if cleaned.strip() != "" else "Unknown"

# Apply text standardization across all object/string columns dynamically
text_cols = df_cleaned.select_dtypes(include=['object']).columns

for col in text_cols:
    df_cleaned[col] = df_cleaned[col].apply(standardize_text)

print("--- Step 2.1: Text Columns Standardized Successfully ---")
```

```python
# Grouping all 28 columns into their precise, definitive datatypes
integer_columns = [
    'rank', 'subscribers', 'video views', 'uploads', 'video_views_rank', 
    'country_rank', 'channel_type_rank', 'video_views_for_the_last_30_days', 
    'subscribers_for_last_30_days', 'created_year'
]

float_columns = [
    'lowest_monthly_earnings', 'highest_monthly_earnings', 
    'lowest_yearly_earnings', 'highest_yearly_earnings',
    'Gross tertiary education enrollment (%)', 'Population', 
    'Unemployment rate', 'Urban_population', 'Latitude', 'Longitude'
]

string_category_columns = [
    'Youtuber', 'category', 'Title', 'Country', 'Abbreviation', 
    'channel_type', 'created_month'
]

# --- 1. Standardizing Integers ---
for col in integer_columns:
    if col in df_cleaned.columns:
        # Convert any logical errors/strings to numeric, fill blank spots with 0
        df_cleaned[col] = pd.to_numeric(df_cleaned[col], errors='coerce').fillna(0).astype('int64')

# --- 2. Standardizing Floats ---
for col in float_columns:
    if col in df_cleaned.columns:
        # Convert to float numbers, keep 0 for unknown structural values
        df_cleaned[col] = pd.to_numeric(df_cleaned[col], errors='coerce').fillna(0.0).astype('float64')

# --- 3. Standardizing Categories/Strings ---
for col in string_category_columns:
    if col in df_cleaned.columns:
        # Enforce uniform string format and strip trailing white spaces
        df_cleaned[col] = df_cleaned[col].astype(str).str.strip()

print("--- Step 2.2: Data Types Mapped & Converted Across All Columns ---")
```

```python
# Print the datatype summary of the full dataset
print("--- FINAL DATATYPE VERIFICATION ---")
print(df_cleaned.info())
```

```python
# 1. Clear any truncation limits for rows and columns
pd.set_option('display.max_rows', None)
pd.set_option('display.max_columns', None)

# 2. Display the dataframe again
df_cleaned
```

```python
# Step 4. REMOVE UNNECESSARY COLUMNS (KEEP 7, DELETE 21)
# We only need 7 columns out of 28 because we will analyze my question and get inghts from these 7 columns and the remaining 21 column is unnecessary,
# if let suppose we will get more question to analyze so we have our raw dataset is on backup for this work we just need these columns.
# Defining the exact 7 columns needed for our project analysis
columns_to_keep = [
    'Youtuber', 
    'subscribers', 
    'video views', 
    'category', 
    'Country', 
    'highest_monthly_earnings', 
    'subscribers_for_last_30_days'
]

# Filtering df_cleaned to keep ONLY these 7 columns. 
# The other 21 columns (like rank, uploads, population, latitude, etc.) are deleted here.
df_cleaned = df_cleaned[columns_to_keep].copy()
```
