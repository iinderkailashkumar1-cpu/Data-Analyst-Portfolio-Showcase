```python
import pandas as pd
from unidecode import unidecode
```

```python
input_file = "F:\Data Analytics Real time Project portfolio\Spotify\Spotify_Dataset.csv"  # Your's original raw file path
output_file = "F:\Data Analytics Real time Project portfolio\Spotify\Spotify_Cleaned_Tracks.csv" # your's Original file location
```

```python
# Using encoding UTF8 will correctly read all world's language perfectly
df = pd.read_csv(input_file, encoding='utf-8')
```

```python
# Using Function to Convert in Englsih Language 
def convert_to_english(text):
    if pd.isna(text):
        return ""
    # unidecode will perfectly convert the text into English Language
    return unidecode(str(text))

print(" Characters are converting in English Language...")
df['artists'] = df['artists'].apply(convert_to_english)
df['album_name'] = df['album_name'].apply(convert_to_english)
df['track_name'] = df['track_name'].apply(convert_to_english)
```

```python
# your folder path where you want to save the file
output_file = "F:\Data Analytics Real time Project portfolio\Spotify\Spotify_Cleaned_Tracks.csv"

# Final save the Clean file
df.to_csv(output_file, index=False, encoding='utf-8')
```
